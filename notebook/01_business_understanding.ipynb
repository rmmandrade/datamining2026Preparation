# ACDC - Customer Segmentation for a Coworking Business
# 1. Business Understanding

## 1.1 Company Context

Amazing Creative Desk Coworking (ACDC) is a coworking operator serving a diverse member base that ranges from freelancers and individual professionals to teams and companies sharing an account. Members choose between several membership tiers and can complement their plan with add-on services such as meeting rooms, event space, guest passes, printing, mail handling, phone booths, day passes, storage and bike lockers.

The coworking industry is highly competitive, and consumers are increasingly selective about where they spend their money. In this context, understanding who the members are and how they use the space is key to improving satisfaction, retention and revenue.

Available data:

- Membership and space-usage activity collected over a three-year period.
- One row per member, combining demographic, contractual, behavioral (visits, add-ons) and financial (lifetime value) information.
- Members include both active and churned customers (a membership end date exists for some members).

This project is carried out as a consulting exercise: the team acts as external consultants to ACDC and must translate data into a segmentation-driven marketing strategy.

## 1.2 The Business Problem

ACDC serves very different types of members under a largely uniform commercial approach. Some members prioritize premium services and dedicated space, while others are cost-conscious and focused on basic desk access. Treating them alike creates three structural problems.

### Problem 1 - One-Size-Fits-All Marketing

Without a clear view of member segments, communication, pricing and perks are not tailored to what each group values. This dilutes marketing spend, weakens the value proposition for each member type and misses upsell opportunities (for example, moving a frequent hot-desk user to a higher tier).

### Problem 2 - Hidden Retention and Value Risks

Members differ widely in how much revenue they generate (lifetime_value), how often they actually show up (actual_visits_last_period vs entitled_visits_last_period, days_since_last_visit) and how loyal they are (n_renewals). Without segmentation, it is hard to tell apart:

- High-value members that deserve protection and recognition.
- Under-using members who pay for more than they use and may be at risk of churning.
- Low-engagement members with little revenue and little activity.

### Problem 3 - No Single View of the Customer

Value, behavior and profile information sit in different variables and tell different stories. A segmentation based on only one perspective (for example, revenue) misses how members actually use the space, and vice versa. ACDC needs an **integrated** view that combines economic contribution, usage habits and other relevant perspectives into one coherent framework that marketing teams can act on.

## 1.3 Project Objective

### Objective - Build a Multi-Perspective Customer Segmentation Framework

The core goal is to use unsupervised learning to divide ACDC's members into smaller groups that share similar characteristics or behaviors, and to turn those groups into a data-driven marketing strategy.

The segmentation will be approached from multiple perspectives:

| Perspective | Question it answers | Example variables |
|:---|:--|:--|
| Value-based | How much does each member contribute economically? | lifetime_value, plan_tier, n_renewals, add-on usage |
| Behavioral | How and how often do members use the space? | total_visits,avg_weekly_visits, days_since_last_visit, usage vs entitlement, meeting room / event space hours |
| Other relevant perspectives | What else differentiates members? | Demographic / firmographic (age, country, industry, is_team_account), tenure and loyalty |

The perspectives are then merged into a final segmentation solution, supporting ACDC in crafting a comprehensive marketing strategy (targeting, service customization and loyalty program optimization).
